```python
import sys
sys.path.append('../src')
from data_utils import download_stock_data
from features import add_features, FEATURE_COLUMNS
```

```python
ticker = 'AAPL'  # Examples: 'RELIANCE.NS', 'TCS.NS'
df = download_stock_data(ticker, period='5y')
df.head()
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt
df['Close'].plot(figsize=(12, 5), title=f'{ticker} Closing Price')
plt.xlabel('Date')
plt.ylabel('Price')
plt.show()
```

```python
data = add_features(df)
data[['Close', 'SMA_20', 'SMA_50', 'EMA_12', 'EMA_26']].plot(figsize=(12, 6), title=f'{ticker} Technical Indicators')
plt.xlabel('Date')
plt.ylabel('Price')
plt.show()
```
